```python
#1.Find the error in the following Python code:
"""
import math
print(Math.sqrt(16))
"""
#first letter of math in print was capital letter
import math
print(math.sqrt(16))
```

```python
#2.Find the error in the following Python code:
"""
def my_function():
    x += 5
    print(x)

x = 10
my_function()
"""

#parameter was not given in the function.
def my_function(x):
    x += 5
    print(x)

x = 10
my_function(x)
```

```python
#3.
"""
x = 5
if x = 10:
    print("x is 10")
"""
#instead of comparison operator used assignment operator
x = 5
if x == 10:
    print("x is 10")
else:
    print("x is not 1")
```

```python
#4.
"""
numbers = [5, 3, 8, 2]
print(numbers.sort())
"""

#need to sort before print

numbers = [5, 3, 8, 2]
numbers.sort()
print(numbers)
```

```python
#5.
"""
def square(n):
    n * n

print(square(4))
"""
#return function was not used
def square(n):
    return n * n

print(square(4))
```

```python
#6.
"""
my_tuple = (1, 2, 3)
my_tuple[1] = 5
"""
#tuple was mutable, so unable to modify 
my_list = [1, 2, 3]
my_list[1] = 5
print(my_list)
```

```python
#7.
"""
name = "Alice"
age = 25
print("My name is %s and I am %d years old" % (name))
"""
#specifier was not mentioned properly
name = "Alice"
age = 25
print("My name is %s and I am %d years old"%(name,age))
```

```python
#8.
"""
a = [1, 2, 3]
b = [1, 2, 3]
if a is b:
    print("Lists are equal")
"""
#to compare the objects need to use comparison operator
a = [1, 2, 3]
b = [1, 2, 3]
if a == b:
    print("Lists are equal")
```

```python
#9.
"""
for i in range(5, 1):
    print(i)
"""
#in the loop start and stop values are placed incorrectly
for i in range(1, 5):
    print(i)
```
