```python
#5.What is wrong with this code?
"""
def is_even(n):
    if n % 2 == 1:
        return True
    else:
        return False
"""

def is_even(n):
    if n % 2 == 0:
        return True
    else:
        return False
is_even(4)

#if condition should compare 0 to check even or not, and function is not called.
```
